```python
import pandas as pd
import chromadb
from sentence_transformers import SentenceTransformer
import anthropic

# ── 1. Load verified data ──────────────────────────────────────────────────
df = pd.read_csv(r"C:\Users\HP\Documents\Python_Courses_and_Projects\Courses\project\stage3_data\simad_ogaal_dataset.csv")
df_verified = df[df['verified'] == 'yes'].reset_index(drop=True)
print(f"✅ Loaded {len(df_verified)} verified Q&A pairs")

# ── 2. Load embedding model ────────────────────────────────────────────────
model = SentenceTransformer('all-MiniLM-L6-v2')

# ── 3. Build vector database ───────────────────────────────────────────────
client_db = chromadb.Client()
try:
    client_db.delete_collection("ogaal_simad")
except:
    pass
collection = client_db.create_collection("ogaal_simad")

collection.add(
    documents=df_verified['question_english'].tolist(),
    metadatas=[{"answer": a, "category": c}
               for a, c in zip(df_verified['answer_english'], df_verified['category'])],
    ids=[str(i) for i in df_verified['id'].tolist()]
)
print(f"✅ {len(df_verified)} questions stored in vector database")

# ── 4. Claude API client ───────────────────────────────────────────────────
claude = anthropic.Anthropic(api_key="sk-ant-***REMOVED***")

# ── 5. RAG + Claude function ───────────────────────────────────────────────
def ask_ogaal(question):
    # Step A: retrieve top 3 relevant Q&A pairs
    results = collection.query(query_texts=[question], n_results=3)
    
    context = ""
    for doc, meta in zip(results['documents'][0], results['metadatas'][0]):
        context += f"Q: {doc}\nA: {meta['answer']}\n\n"
    
    # Step B: send context + question to Claude
    prompt = f"""You are Ogaal, the official assistant of SIMAD AI Institute in Mogadishu, Somalia.
Use only the information below to answer the question. Be clear and helpful.

--- INFORMATION ---
{context}
--- END ---

Question: {question}
Answer:"""

    response = claude.messages.create(
        model="claude-haiku-4-5",
        max_tokens=300,
        messages=[{"role": "user", "content": prompt}]
    )
    
    return response.content[0].text

# ── 6. Test it ─────────────────────────────────────────────────────────────
questions = [
    "How can I join SIMAD?",
    "What documents do I need to apply?",
    "Who can study at SIMAD AI Institute?"
]

for q in questions:
    print(f"\n🙋 {q}")
    print(f"🤖 {ask_ogaal(q)}")
    print("-" * 60)
```

```python
# ── INSTALL PACKAGES ──────────────────────────────────────────────────────
import sys
!{sys.executable} -m pip install chromadb sentence-transformers anthropic python-dotenv flask flask-cors -q

# ── IMPORTS ───────────────────────────────────────────────────────────────
import os
import pandas as pd
import chromadb
from sentence_transformers import SentenceTransformer
import anthropic
from dotenv import load_dotenv
from flask import Flask, request, jsonify, Response
from flask_cors import CORS
import threading

# ── LOAD API KEY ───────────────────────────────────────────────────────────
load_dotenv(r"C:\Users\HP\Documents\Python_Courses_and_Projects\Courses\project\.env")
ANTHROPIC_API_KEY = os.getenv("ANTHROPIC_API_KEY")

# ── LOAD DATASET ───────────────────────────────────────────────────────────
df = pd.read_csv(r"C:\Users\HP\Documents\Python_Courses_and_Projects\Courses\project\stage3_data\simad_ogaal_dataset.csv")
df_verified = df[df['verified'] == 'yes'].reset_index(drop=True)
print(f"✅ Loaded {len(df_verified)} verified Q&A pairs")

# ── EMBEDDING MODEL ────────────────────────────────────────────────────────
print("Loading embedding model...")
model = SentenceTransformer('all-MiniLM-L6-v2')
print("✅ Model ready")

# ── VECTOR DATABASE ────────────────────────────────────────────────────────
client_db = chromadb.Client()
try:
    client_db.delete_collection("ogaal_simad")
except:
    pass
collection = client_db.create_collection("ogaal_simad")
collection.add(
    documents=df_verified['question_english'].tolist(),
    metadatas=[{"answer": a, "category": c}
               for a, c in zip(df_verified['answer_english'], df_verified['category'])],
    ids=[str(i) for i in df_verified['id'].tolist()]
)
print(f"✅ {len(df_verified)} questions stored in vector database")

# ── CLAUDE CLIENT ──────────────────────────────────────────────────────────
claude = anthropic.Anthropic(api_key=ANTHROPIC_API_KEY)

# ── RAG FUNCTION ───────────────────────────────────────────────────────────
def ask_ogaal(question):
    results = collection.query(query_texts=[question], n_results=3)
    context = ""
    for doc, meta in zip(results['documents'][0], results['metadatas'][0]):
        context += f"Q: {doc}\nA: {meta['answer']}\n\n"
    prompt = f"""You are Ogaal, the official assistant of SIMAD AI Institute in Mogadishu, Somalia.
Use only the information below to answer the question. Be clear and helpful.

--- INFORMATION ---
{context}
--- END ---

Question: {question}
Answer:"""
    response = claude.messages.create(
        model="claude-haiku-4-5",
        max_tokens=300,
        messages=[{"role": "user", "content": prompt}]
    )
    return response.content[0].text

# ── CHAT INTERFACE HTML ────────────────────────────────────────────────────
HTML = """<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>Ogaal - SIMAD AI Institute</title>
<script src="https://cdnjs.cloudflare.com/ajax/libs/marked/9.1.6/marked.min.js"></script>
<style>
  * { margin: 0; padding: 0; box-sizing: border-box; }
  body { font-family: 'Segoe UI', sans-serif; background: #f0f2f5; height: 100vh; display: flex; flex-direction: column; align-items: center; justify-content: center; }
  .chat-container { width: 100%; max-width: 700px; height: 90vh; background: white; border-radius: 16px; box-shadow: 0 4px 20px rgba(0,0,0,0.1); display: flex; flex-direction: column; overflow: hidden; }
  .header { background: #1a4f8a; color: white; padding: 20px 24px; }
  .header h1 { font-size: 20px; font-weight: 600; }
  .header p { font-size: 13px; opacity: 0.8; margin-top: 4px; }
  .messages { flex: 1; overflow-y: auto; padding: 20px; display: flex; flex-direction: column; gap: 12px; }
  .message { max-width: 80%; padding: 12px 16px; border-radius: 12px; font-size: 14px; line-height: 1.6; }
  .user { align-self: flex-end; background: #1a4f8a; color: white; border-bottom-right-radius: 4px; }
  .bot { align-self: flex-start; background: #f0f2f5; color: #1a1a1a; border-bottom-left-radius: 4px; }
  .bot .name { font-size: 11px; font-weight: 600; color: #1a4f8a; margin-bottom: 6px; }
  .bot p { margin-bottom: 6px; }
  .bot strong { font-weight: 600; }
  .bot h1, .bot h2, .bot h3 { font-size: 14px; font-weight: 600; margin-bottom: 6px; }
  .bot ul, .bot ol { padding-left: 18px; margin-bottom: 6px; }
  .input-area { padding: 16px 20px; border-top: 1px solid #e5e7eb; display: flex; gap: 10px; }
  input { flex: 1; padding: 12px 16px; border: 1px solid #d1d5db; border-radius: 24px; font-size: 14px; outline: none; }
  input:focus { border-color: #1a4f8a; }
  button { background: #1a4f8a; color: white; border: none; padding: 12px 20px; border-radius: 24px; font-size: 14px; cursor: pointer; font-weight: 500; }
  button:hover { background: #15407a; }
  .welcome { text-align: center; color: #6b7280; font-size: 13px; margin: auto; }
</style>
</head>
<body>
<div class="chat-container">
  <div class="header">
    <h1>🎓 Ogaal Assistant</h1>
    <p>SIMAD AI Institute — Ask me anything about admissions</p>
  </div>
  <div class="messages" id="messages">
    <div class="welcome">
      <p>👋 Hello! I am Ogaal, the SIMAD AI Institute assistant.</p>
      <p style="margin-top:8px;">Ask me about admissions, requirements, or how to apply.</p>
    </div>
  </div>
  <div class="input-area">
    <input type="text" id="userInput" placeholder="Type your question..." onkeypress="if(event.key==='Enter') sendMessage()" />
    <button onclick="sendMessage()">Send</button>
  </div>
</div>
<script>
async function sendMessage() {
  const input = document.getElementById('userInput');
  const messages = document.getElementById('messages');
  const question = input.value.trim();
  if (!question) return;
  const welcome = messages.querySelector('.welcome');
  if (welcome) welcome.remove();
  messages.innerHTML += `<div class="message user">${question}</div>`;
  input.value = '';
  messages.innerHTML += `<div class="message bot" id="typing"><div class="name">Ogaal</div>Thinking...</div>`;
  messages.scrollTop = messages.scrollHeight;
  try {
    const res = await fetch('/ask', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      body: JSON.stringify({ question })
    });
    const data = await res.json();
    document.getElementById('typing').remove();
    messages.innerHTML += `<div class="message bot"><div class="name">Ogaal</div>${marked.parse(data.answer)}</div>`;
  } catch (e) {
    document.getElementById('typing').remove();
    messages.innerHTML += `<div class="message bot"><div class="name">Ogaal</div>Sorry, could not connect to the server.</div>`;
  }
  messages.scrollTop = messages.scrollHeight;
}
</script>
</body>
</html>"""

# ── FLASK SERVER ───────────────────────────────────────────────────────────
app = Flask(__name__)
CORS(app)

@app.route('/')
def index():
    return Response(HTML, mimetype='text/html')

@app.route('/ask', methods=['POST'])
def ask():
    data = request.get_json()
    answer = ask_ogaal(data.get('question', ''))
    return jsonify({"answer": answer})

def run_server():
    app.run(port=5001, debug=False, use_reloader=False)

thread = threading.Thread(target=run_server)
thread.daemon = True
thread.start()

print("\n✅ Ogaal chatbot is ready!")
print("➡️  Open in Chrome: http://localhost:5001")
```
